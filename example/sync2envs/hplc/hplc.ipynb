```python
import matplotlib.pyplot as plt
import numpy as np


def plot_selection_result(results, ax, budget=None, label=None, color=None):
    import scipy.stats as sps

    if budget is None:
        budget = results.shape[-1]

    def ci(y):
        br = sps.bootstrap((y,), np.mean, vectorized=True)
        ci_l = br.confidence_interval.low
        ci_u = br.confidence_interval.high
        return ci_l, ci_u

    iters = np.arange(0, budget, 1)
    agg_value = results.mean(axis=0)[:budget].ravel()
    _ci_lb, _ci_ub = ci(results)[:budget]

    line, *_ = ax.plot(iters, agg_value, linewidth=1.5, label=label, color=color)

    ax.plot(iters, _ci_lb, linewidth=1.5, alpha=0.5, color=line.get_color())
    ax.plot(iters, _ci_ub, linewidth=1.5, alpha=0.5, color=line.get_color())
    ax.fill_between(iters, _ci_lb, _ci_ub, alpha=0.3, color=line.get_color())
```

```python
import numpy as np
import torch

# random res
random_out = np.loadtxt("random/random_out.txt")
random_out = random_out.reshape(-1, 100)
random_res = torch.from_numpy(random_out).cummax(axis=1).values.detach().numpy()

# optuna 
optuna_out = np.loadtxt("optuna/optuna_values.txt")
optuna_res = torch.from_numpy(optuna_out).cummax(axis=1).values.detach().numpy()

# hebo
hebo_out = np.loadtxt("hebo/hebo_values.txt")
hebo_res = torch.from_numpy(hebo_out).cummax(axis=1).values.detach().numpy()

# tidyhebo
tidyhebo_out = np.loadtxt("tidyhebo/tidyhebo_values.txt")
tidyhebo_res = torch.from_numpy(tidyhebo_out).cummax(axis=1).values.detach().numpy()

```

```python
import matplotlib.pyplot as plt

ax = plt.gca()

data_known_max = 2569.87964
# optimization_got_max = max(random_res.max(), optuna_res.max(), hebo_res.max(), tidyhebo_res.max())

tr = lambda x: x
# tr = lambda x: np.log10(optimization_got_max - x)

plot_selection_result(tr(random_res), ax, budget=100, label="random", color="tab:blue")
plot_selection_result(tr(optuna_res), ax, budget=100, label="optuna", color="tab:green")
plot_selection_result(tr(hebo_res), ax, budget=100, label="hebo", color="tab:purple")
plot_selection_result(tr(tidyhebo_res), ax, budget=100, label="tidyhebo", color="tab:red")


ax.legend()
```

```python
import matplotlib.pyplot as plt

ax = plt.gca()

data_known_max = 2569.87964
optimization_got_max = max(random_res.max(), optuna_res.max(), hebo_res.max(), tidyhebo_res.max())

# tr = lambda x: x
tr = lambda x: np.log10(optimization_got_max - x)

plot_selection_result(tr(random_res), ax, budget=100, label="random", color="tab:blue")
plot_selection_result(tr(optuna_res), ax, budget=100, label="optuna", color="tab:green")
plot_selection_result(tr(hebo_res), ax, budget=100, label="hebo", color="tab:purple")
plot_selection_result(tr(tidyhebo_res), ax, budget=100, label="tidyhebo", color="tab:red")


ax.legend()
```

```python
import matplotlib.pyplot as plt


fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(11, 5.2))

optimization_got_max = max(random_res.max(), optuna_res.max(), hebo_res.max(), tidyhebo_res.max())

def log_regret(x):
    return np.log10(optimization_got_max * 1.01 - np.array(x))


plot_selection_result(random_res, ax1, budget=100, label="random", color="tab:blue")
plot_selection_result(optuna_res, ax1, budget=100, label="optuna", color="tab:green")
plot_selection_result(hebo_res, ax1, budget=100, label="hebo", color="tab:purple")
plot_selection_result(tidyhebo_res, ax1, budget=100, label="tidyhebo", color="tab:red")


ax1.legend()
fig.suptitle("Сходимость моделей (hplc)")
ax1.set_ylabel("best value")
ax1.set_xlabel("# f-evals")
ax1.set_ylim(bottom=1200)

plot_selection_result(log_regret(random_res), ax2, budget=100, label="random", color="tab:blue")
plot_selection_result(log_regret(optuna_res), ax2, budget=100, label="optuna", color="tab:green")
plot_selection_result(log_regret(hebo_res), ax2, budget=100, label="hebo", color="tab:purple")
plot_selection_result(log_regret(tidyhebo_res), ax2, budget=100, label="tidyhebo", color="tab:red")


ax2.set_ylabel("log10 regret")
ax2.set_xlabel("# f-evals")

fig.tight_layout()
```
