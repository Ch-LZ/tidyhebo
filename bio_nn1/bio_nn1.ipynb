```python
import pandas as pd
import torch


df = pd.read_csv("y_from_dbp.csv", index_col=0)
data = torch.from_numpy(df.values.ravel()).cummax(dim=0).values.detach().numpy()
```

```python
import matplotlib.pyplot as plt


plt.plot(data)
```

```python
y = df.values.ravel()
x = list(range(len(y)))

plt.scatter(x=x, y=y, label="val", alpha=0.8, color="tab:green")
ax = plt.gca()

ax.set_ylim(bottom=0.955, top=0.975)

ax.set_ylabel("validation loss")
ax.set_xlabel("# f-evals")

ax.plot(data, label="maxval", color="tab:green")
ax.legend()
ax.set_title("deep neural network")
```
