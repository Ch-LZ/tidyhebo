# tidyHEBO paper presets

This example uses the retained qMACE configuration with a batch size of four. The presets configure the optimizer. Experiment tasks, initialization protocols and repeat seeds belong to the separate results repository.


```python
from dataclasses import replace
import random

import numpy as np
import torch

from tidyhebo import RealP, tidyHEBO
from tidyhebo.presets import tidyHEBO_QMACE

seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
torch.set_num_threads(1)

config = replace(tidyHEBO_QMACE, acquisition=replace(tidyHEBO_QMACE.acquisition, qmc_seed=seed))
optimizer = tidyHEBO.from_config(config, [RealP(0, 0.0, 1.0, name="x")], n_init=4)

for _ in range(2):
    X = optimizer.suggest()
    y = -(X[["x"]].to_numpy() - 0.35) ** 2
    optimizer.observe(X, y)

print("Evaluations:", len(optimizer.train_y))
print("Best observed objective:", optimizer.train_y.max().item())
```

The comparison presets are `tidyHEBO_QLOGEI` for the warped GP and `VANILLA_GP_QLOGEI` for the unwarped GP. To use a batch size of eight, set `q=8` in the inner `replace` call.
